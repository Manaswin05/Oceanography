```python
# =====================================================================
# FISH IMAGE CLASSIFICATION — COMPLETE COLAB NOTEBOOK
# 9 classes: Sea Bass, Trout, Striped Red Mullet, Red Sea Bream, Shrimp,
# Red Mullet, Hourse Mackerel, Gilt-Head Bream, Black Sea Sprat
# =====================================================================
# Copy each "STAGE" block into its own Colab cell (or run as one script).

# =====================================================================
# STAGE 0: INSTALL / IMPORTS
# =====================================================================
import os
import shutil
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix
from google.colab import drive

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

print("⚙️ Stage 0 Complete: Libraries imported & random seeds locked.")

# =====================================================================
# STAGE 1: DATA INGESTION & LOCAL STORAGE CACHING
# =====================================================================
drive.mount('/content/drive')

def locate_dataset_on_drive(start_path='/content/drive/MyDrive'):
    print("🔍 Searching Google Drive for 'Fish_Dataset'...")
    for root, dirs, files in os.walk(start_path):
        if 'Sea Bass' in dirs or 'Red Sea Bream' in dirs or 'Trout' in dirs:
            return root
    return None

source_dir = locate_dataset_on_drive()
clean_dir = '/content/Clean_Fish_Dataset'

if source_dir is None:
    raise FileNotFoundError("❌ Could not locate 'Fish_Dataset' on Google Drive. Please verify upload.")

print(f"✅ Located dataset at: {source_dir}")

# =====================================================================
# STAGE 2: DATA SANITIZATION & CLEANING
# =====================================================================
os.makedirs(clean_dir, exist_ok=True)
species_count = 0

print("📦 Copying and sanitizing dataset (filtering out GT mask folders & hidden files)...")
for folder_name in os.listdir(source_dir):
    folder_path = os.path.join(source_dir, folder_name)

    # Filter: Keep valid directories, drop 'GT' mask folders & hidden OS files (.DS_Store)
    if os.path.isdir(folder_path) and not folder_name.upper().endswith('GT') and not folder_name.startswith('.'):
        target_path = os.path.join(clean_dir, folder_name)
        if not os.path.exists(target_path):
            shutil.copytree(folder_path, target_path)
        print(f"  --> Sanitized Class: {folder_name}")
        species_count += 1

print(f"\n🎉 Stage 2 Complete: Extracted {species_count} clean species classes into local storage.")

# =====================================================================
# STAGE 3: BUILD FILE INDEX & CHECK CLASS BALANCE
# =====================================================================
CLASS_NAMES = sorted(os.listdir(clean_dir))
print(f"Classes found ({len(CLASS_NAMES)}): {CLASS_NAMES}")

VALID_EXT = ('.png', '.jpg', '.jpeg', '.bmp')
records = []
for cls in CLASS_NAMES:
    cls_dir = os.path.join(clean_dir, cls)
    # Walk recursively: this dataset nests images one level deeper
    # e.g. "Sea Bass/Sea Bass/*.png" and "Sea Bass/Sea Bass GT/*.png"
    for root, dirs, files in os.walk(cls_dir):
        # Skip any leftover GT (ground-truth mask) subfolders
        dirs[:] = [d for d in dirs if not d.upper().endswith('GT')]
        if os.path.basename(root).upper().endswith('GT'):
            continue
        for fname in files:
            if fname.lower().endswith(VALID_EXT):
                records.append({'filepath': os.path.join(root, fname), 'label': cls})

df = pd.DataFrame(records)
print(f"Total usable images: {len(df)}")
if len(df) == 0:
    raise ValueError(
        "❌ No images found. Inspect the folder structure with "
        "os.walk(clean_dir) to confirm where image files actually live."
    )
print(df['label'].value_counts())

plt.figure(figsize=(10, 4))
sns.countplot(data=df, y='label', order=df['label'].value_counts().index)
plt.title("Class Distribution")
plt.xlabel("Image Count")
plt.tight_layout()
plt.show()

# =====================================================================
# STAGE 4: TRAIN / VALIDATION / TEST SPLIT (stratified, 70/15/15)
# =====================================================================
train_df, temp_df = train_test_split(
    df, test_size=0.30, stratify=df['label'], random_state=SEED
)
val_df, test_df = train_test_split(
    temp_df, test_size=0.50, stratify=temp_df['label'], random_state=SEED
)

print(f"Train: {len(train_df)} | Val: {len(val_df)} | Test: {len(test_df)}")

# =====================================================================
# STAGE 5: tf.data PIPELINES (with augmentation on train only)
# =====================================================================
IMG_SIZE = (224, 224)
BATCH_SIZE = 32
NUM_CLASSES = len(CLASS_NAMES)
class_to_idx = {c: i for i, c in enumerate(CLASS_NAMES)}

def load_and_preprocess(filepath, label, augment=False):
    img = tf.io.read_file(filepath)
    img = tf.image.decode_image(img, channels=3, expand_animations=False)
    img = tf.image.resize(img, IMG_SIZE)

    if augment:
        img = tf.image.random_flip_left_right(img)
        img = tf.image.random_brightness(img, max_delta=0.15)
        img = tf.image.random_contrast(img, lower=0.85, upper=1.15)
        img = tf.image.rot90(img, k=random.choice([0, 1, 2, 3]))

    img = preprocess_input(img)  # MobileNetV2-specific scaling
    return img, label

def make_dataset(dframe, augment=False, shuffle=False):
    filepaths = dframe['filepath'].values
    labels = dframe['label'].map(class_to_idx).values
    labels = tf.keras.utils.to_categorical(labels, num_classes=NUM_CLASSES)

    ds = tf.data.Dataset.from_tensor_slices((filepaths, labels))
    if shuffle:
        ds = ds.shuffle(buffer_size=len(dframe), seed=SEED)
    ds = ds.map(lambda x, y: load_and_preprocess(x, y, augment=augment),
                num_parallel_calls=tf.data.AUTOTUNE)
    ds = ds.batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)
    return ds

train_ds = make_dataset(train_df, augment=True, shuffle=True)
val_ds = make_dataset(val_df, augment=False, shuffle=False)
test_ds = make_dataset(test_df, augment=False, shuffle=False)

print("⚙️ Stage 5 Complete: tf.data pipelines built.")

# =====================================================================
# STAGE 6: MODEL — TRANSFER LEARNING WITH MOBILENETV2
# =====================================================================
base_model = MobileNetV2(
    input_shape=IMG_SIZE + (3,),
    include_top=False,
    weights='imagenet'
)
base_model.trainable = False  # freeze for initial training phase

inputs = layers.Input(shape=IMG_SIZE + (3,))
x = base_model(inputs, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.3)(x)
x = layers.Dense(128, activation='relu')(x)
x = layers.Dropout(0.2)(x)
outputs = layers.Dense(NUM_CLASSES, activation='softmax')(x)

model = models.Model(inputs, outputs)

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()

# =====================================================================
# STAGE 7: CALLBACKS
# =====================================================================
os.makedirs('/content/checkpoints', exist_ok=True)

callbacks = [
    tf.keras.callbacks.EarlyStopping(
        monitor='val_loss', patience=5, restore_best_weights=True
    ),
    tf.keras.callbacks.ReduceLROnPlateau(
        monitor='val_loss', factor=0.5, patience=2, min_lr=1e-6
    ),
    tf.keras.callbacks.ModelCheckpoint(
        '/content/checkpoints/best_model.keras',
        monitor='val_accuracy', save_best_only=True
    )
]

# =====================================================================
# STAGE 8: PHASE 1 TRAINING — FROZEN BASE
# =====================================================================
EPOCHS_PHASE1 = 15

history1 = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS_PHASE1,
    callbacks=callbacks
)

# =====================================================================
# STAGE 9: PHASE 2 — FINE-TUNING (unfreeze top layers of base model)
# =====================================================================
base_model.trainable = True
FINE_TUNE_AT = len(base_model.layers) - 40  # unfreeze last 40 layers

for layer in base_model.layers[:FINE_TUNE_AT]:
    layer.trainable = False

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-5),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

EPOCHS_PHASE2 = 10

history2 = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS_PHASE2,
    callbacks=callbacks
)

# =====================================================================
# STAGE 10: TRAINING CURVES
# =====================================================================
def combine_histories(h1, h2, key):
    return h1.history[key] + h2.history[key]

acc = combine_histories(history1, history2, 'accuracy')
val_acc = combine_histories(history1, history2, 'val_accuracy')
loss = combine_histories(history1, history2, 'loss')
val_loss = combine_histories(history1, history2, 'val_loss')

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].plot(acc, label='Train Accuracy')
axes[0].plot(val_acc, label='Val Accuracy')
axes[0].axvline(x=len(history1.history['accuracy']) - 1, color='gray', linestyle='--', label='Fine-tune start')
axes[0].set_title('Accuracy over Epochs')
axes[0].set_xlabel('Epoch')
axes[0].legend()

axes[1].plot(loss, label='Train Loss')
axes[1].plot(val_loss, label='Val Loss')
axes[1].axvline(x=len(history1.history['loss']) - 1, color='gray', linestyle='--', label='Fine-tune start')
axes[1].set_title('Loss over Epochs')
axes[1].set_xlabel('Epoch')
axes[1].legend()

plt.tight_layout()
plt.show()

# =====================================================================
# STAGE 11: EVALUATION ON TEST SET
# =====================================================================
test_loss, test_acc = model.evaluate(test_ds)
print(f"\n✅ Test Accuracy: {test_acc:.4f} | Test Loss: {test_loss:.4f}")

y_true = []
y_pred = []
for images, labels in test_ds:
    preds = model.predict(images, verbose=0)
    y_true.extend(np.argmax(labels.numpy(), axis=1))
    y_pred.extend(np.argmax(preds, axis=1))

print("\nClassification Report:")
print(classification_report(y_true, y_pred, target_names=CLASS_NAMES))

cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=CLASS_NAMES, yticklabels=CLASS_NAMES)
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# =====================================================================
# STAGE 12: SAVE FINAL MODEL (to Drive for persistence)
# =====================================================================
save_path = '/content/drive/MyDrive/fish_classifier_mobilenetv2.keras'
model.save(save_path)
print(f"💾 Model saved to: {save_path}")

# Also save class name mapping for inference
import json
with open('/content/drive/MyDrive/fish_class_names.json', 'w') as f:
    json.dump(CLASS_NAMES, f)
print("💾 Class names saved to: /content/drive/MyDrive/fish_class_names.json")

# =====================================================================
# STAGE 13: INFERENCE HELPER — PREDICT ON A NEW IMAGE
# =====================================================================
def predict_fish(image_path, model=model, class_names=CLASS_NAMES):
    img = tf.io.read_file(image_path)
    img = tf.image.decode_image(img, channels=3, expand_animations=False)
    img = tf.image.resize(img, IMG_SIZE)
    img = preprocess_input(img)
    img = tf.expand_dims(img, axis=0)

    preds = model.predict(img, verbose=0)[0]
    top_idx = np.argmax(preds)

    plt.imshow(tf.io.decode_image(tf.io.read_file(image_path)))
    plt.axis('off')
    plt.title(f"Predicted: {class_names[top_idx]} ({preds[top_idx]*100:.1f}%)")
    plt.show()

    print("Top 3 predictions:")
    top3 = np.argsort(preds)[-3:][::-1]
    for i in top3:
        print(f"  {class_names[i]}: {preds[i]*100:.2f}%")

    return class_names[top_idx], preds[top_idx]

# Example usage (uncomment and set a real path):
# predict_fish('/content/Clean_Fish_Dataset/Trout/some_image.png')

print("\n🎉 ALL STAGES COMPLETE — model trained, evaluated, and saved.")
```

```python
# =====================================================================
# TEST SAVED MODEL ON UPLOADED IMAGES (run in a fresh or existing Colab session)
# =====================================================================
import os
import json
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input
from google.colab import drive, files

# ---------------------------------------------------------------------
# STEP 1: Mount Drive & load the saved model + class names
# ---------------------------------------------------------------------
drive.mount('/content/drive')

MODEL_PATH = '/content/drive/MyDrive/fish_classifier_mobilenetv2.keras'
CLASS_NAMES_PATH = '/content/drive/MyDrive/fish_class_names.json'
IMG_SIZE = (224, 224)

model = tf.keras.models.load_model(MODEL_PATH)
with open(CLASS_NAMES_PATH, 'r') as f:
    CLASS_NAMES = json.load(f)

print(f"✅ Model loaded. Classes: {CLASS_NAMES}")

# ---------------------------------------------------------------------
# STEP 2: Upload one or more images from your computer
# ---------------------------------------------------------------------
print("📤 Select one or more fish images to upload...")
uploaded = files.upload()  # opens a file picker; select any number of images
uploaded_paths = list(uploaded.keys())
print(f"Uploaded {len(uploaded_paths)} image(s): {uploaded_paths}")

# ---------------------------------------------------------------------
# STEP 3: Prediction helper
# ---------------------------------------------------------------------
def predict_fish(image_path, model=model, class_names=CLASS_NAMES, top_k=3):
    raw = tf.io.read_file(image_path)
    img = tf.image.decode_image(raw, channels=3, expand_animations=False)
    img_resized = tf.image.resize(img, IMG_SIZE)
    img_input = preprocess_input(img_resized)
    img_input = tf.expand_dims(img_input, axis=0)

    preds = model.predict(img_input, verbose=0)[0]
    top_idx = np.argmax(preds)

    plt.figure(figsize=(4, 4))
    plt.imshow(img.numpy().astype('uint8'))
    plt.axis('off')
    plt.title(f"{os.path.basename(image_path)}\n"
              f"Predicted: {class_names[top_idx]} ({preds[top_idx]*100:.1f}%)")
    plt.show()

    top_k_idx = np.argsort(preds)[-top_k:][::-1]
    print("Top predictions:")
    for i in top_k_idx:
        print(f"  {class_names[i]:<20s} {preds[i]*100:5.2f}%")
    print()

    return class_names[top_idx], float(preds[top_idx])

# ---------------------------------------------------------------------
# STEP 4: Run predictions on every uploaded image
# ---------------------------------------------------------------------
results = {}
for path in uploaded_paths:
    label, confidence = predict_fish(path)
    results[path] = (label, confidence)

print("Summary:")
for path, (label, conf) in results.items():
    print(f"  {path} -> {label} ({conf*100:.2f}%)")
```
